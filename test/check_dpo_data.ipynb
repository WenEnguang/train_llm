```python
# 检查DPO数据质量，看是否存在chosen/rejected的区分度不高，导致的模型效果不高
from pathlib import Path
import os
from rich import print as rprint
import json

root_dir  = Path().resolve().parent
raw_data_dir = root_dir / 'data' / 'raw' 

rprint(f"root dir:{raw_data_dir}")

file_path = raw_data_dir / 'dpo.jsonl'

# 打印当前文件的部分数据
with open(file_path, 'r', encoding='utf-8') as f:
    for i, line in enumerate(f,start=6):
        if i > 11:
            break
        line = line.strip()
        if line :
            rprint(f"[cyan]--- 第 {i+1} 条 ---[/cyan]")
            try:
                data = json.loads(line)
                rprint(data)
            except json.JSONDecodeError as e:
                rprint(f"[red]JSON 解析错误: {e}[/red]")
```
