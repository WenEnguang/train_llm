# 文件大概有多少行/多少GB？决定是否是流式迭代还是全内存加载

```python
import os
import sys
from pathlib import Path

current_dir = os.getcwd()
sys.path.insert(0, current_dir)

import json
from rich import print as rprint

pretrain_file_path = '/home/user/data/2025/wen/train_llm/data/raw/pretrain_t2t_mini.jsonl'

if os.path.exists(pretrain_file_path):
    # 1. 统计文件大小
    p = Path(pretrain_file_path)
    size_bytes = p.stat().st_size
    size_mb = size_bytes / 1024 / 1024
    size_gb = size_mb / 1024

    # 2. 逐行统计行数（最稳妥，超大文件不爆内存）
    line_count = 0
    sample_data = []  # 存前3条样本用来查看格式
    with open(pretrain_file_path, 'r', encoding='utf-8') as f:
        for idx, line in enumerate(f):
            line_count += 1
            # 只取前3行做样例
            if idx < 3:
                try:
                    obj = json.loads(line.strip())
                    sample_data.append(obj)
                except Exception as e:
                    rprint(f"第{idx+1}行解析异常: {e}")

    # 3. 输出结果
    rprint("[green]===== JSONL 文件统计结果 =====[/green]")
    rprint(f"文件路径: {pretrain_file_path}")
    rprint(f"文件大小: {size_bytes:,} Bytes | {size_mb:.2f} MB | {size_gb:.2f} GB")
    rprint(f"总样本行数: {line_count:,}")

    rprint("\n[blue]前3条数据样例：[/blue]")
    for i, item in enumerate(sample_data, 1):
        rprint(f"--- 第{i}条 ---")
        rprint(item)

else:
    rprint(f"[red]文件不存在：{pretrain_file_path}[/red]")
```
